```python
import fastf1
fastf1.Cache.enable_cache('C:\Coding Projects\F1-Data-Analysis\cache')

session = fastf1.get_session(2025, 'Monaco', 'R')
session.load()
```

Extracts high-level metadata for a Formula 1 session (event, location, date, session type).

```python
from datetime import date, datetime

event = session.event

event_date = event.get("EventDate")
if isinstance(event_date, datetime):
    event_date = event_date.date()
elif not isinstance(event_date, date):
    event_date = None

overview = {
    "Season": event.get("Season"),
    "Round": event.get("RoundNumber"),
    "EventName": event.get("EventName"),
    "Location": event.get("Location"),
    "Country": event.get("Country"),
    "Date": event_date,
    "Session": session.name,
}

overview
```
